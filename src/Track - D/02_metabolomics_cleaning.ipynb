```python
import numpy as np
import pandas as pd

def audit(df: pd.DataFrame) -> pd.DataFrame:
    n = len(df)
    miss = df.isna().sum()
    pct = (miss.to_numpy(dtype=float) / n * 100).round(1) if n else np.zeros(len(miss))
    return (pd.DataFrame({
        "column":    miss.index,
        "dtype":     df.dtypes.astype(str).to_numpy(),
        "missing_n": miss.to_numpy(),
        "missing_%": pct,
    })
    .sort_values("missing_%", ascending=False)
    .reset_index(drop=True))
```

```python
import pandas as pd

# --- 1. Isolate the missing-id row ---
miss = metabolomics[metabolomics["depmap_id"].isna()]
print(f"missing rows: {len(miss)}  |  index: {miss.index.tolist()}")
```

```python
# --- Map oc315_ovary -> depmap_id via sample_info ---
id_col = next(c for c in sample_info.columns if c in ("depmap_id"))
sname  = next(c for c in sample_info.columns if "ccle" in c and "name" in c)

key = metabolomics.loc[553, "ccle_id"].strip().upper()   # 'OC315_OVARY'
hit = sample_info.loc[sample_info[sname].str.strip().str.upper() == key, id_col]
print(hit)
```

```python
before = metabolomics["depmap_id"].notna().sum()
metabolomics.loc[553, "depmap_id"] = hit.iloc[0]
after = metabolomics["depmap_id"].notna().sum()
assert after == before + 1
print(f"filled row 553 -> {hit.iloc[0]}  ({before} -> {after})")
```

```python
# --- Build the two name->id maps ---
id_col = next(c for c in sample_info.columns if c in ("depmap_id","modelid","model_id"))
sname  = next(c for c in sample_info.columns if "ccle" in c and "name" in c)

meta_pairs = (metabolomics[["ccle_id", "depmap_id"]]
              .assign(ccle_key=lambda d: d["ccle_id"].str.strip().str.upper()))

si_pairs = (sample_info[[sname, id_col]]
            .rename(columns={sname: "ccle_id_si", id_col: "depmap_id_si"})
            .assign(ccle_key=lambda d: d["ccle_id_si"].str.strip().str.upper()))

# --- Merge on the ccle name and compare the two depmap_ids ---
chk = meta_pairs.merge(si_pairs, on="ccle_key", how="left")

# how many metabolomics names even resolve in sample_info?
unmatched = chk["depmap_id_si"].isna().sum()
print(f"names not found in sample_info: {unmatched} / {len(chk)}")

# where metabolomics already had an id, does it agree with sample_info?
both = chk[chk["depmap_id"].notna() & chk["depmap_id_si"].notna()]
mismatch = both[both["depmap_id"] != both["depmap_id_si"]]
print(f"rows with both ids: {len(both)}  |  disagreements: {len(mismatch)}")
if len(mismatch):
    print(mismatch[["ccle_id", "depmap_id", "depmap_id_si"]].to_string())

# --- The row we care about: 553 (oc315_ovary) ---
print("\nrow 553 resolution:")
print(chk.loc[553, ["ccle_id", "depmap_id", "depmap_id_si"]])
```
