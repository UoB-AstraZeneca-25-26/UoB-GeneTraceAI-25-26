```python
import pandas as pd

def lower_all(df):
    """
    Lowercase column names and all string values in a DataFrame.
    Non-string columns (numeric, bool, datetime) are left untouched.
    Returns a new DataFrame; does not modify in place.
    """
    df = df.copy()

    # lowercase column names
    df.columns = [str(c).lower() for c in df.columns]

    # lowercase string values only, column by column
    for col in df.columns:
        if pd.api.types.is_object_dtype(df[col]) or pd.api.types.is_string_dtype(df[col]):
            df[col] = df[col].str.lower()

    return df
```

```python
def missing_report(df, name=None):
    """Percentage missing per column, sorted worst-first. Shows only cols with >0 missing."""
    n = len(df)
    miss = df.isna().sum()
    pct = (miss / n * 100).round(2)
    out = (
        pd.DataFrame({"n_missing": miss, "pct_missing": pct})
        .query("n_missing > 0")
        .sort_values("pct_missing", ascending=False)
    )
    header = f"{name} — {n:,} rows" if name else f"{n:,} rows"
    print(f"{header}: {len(out)}/{df.shape[1]} cols have missing values")
    return out

# single table
missing_report(cellosaurus, "cellosaurus")
```

```python
def drop_high_missing(df, threshold=80.0, name=None, verbose=True):
    """Drop columns whose missing % strictly exceeds `threshold`. Returns a new DataFrame."""
    n = len(df)
    pct = df.isna().sum() / n * 100
    to_drop = pct[pct > threshold].sort_values(ascending=False)

    if verbose:
        tag = f"{name}: " if name else ""
        print(f"{tag}{df.shape[1]} cols -> dropping {len(to_drop)} above {threshold}% missing")
        for col, p in to_drop.items():
            print(f"    {p:6.2f}%  {col}")

    return df.drop(columns=to_drop.index)

cellosaurus = drop_high_missing(cellosaurus, threshold=80.0, name="cellosaurus")
tables["Cellosaurus"] = cellosaurus   # re-point so the dict doesn't diverge
```
