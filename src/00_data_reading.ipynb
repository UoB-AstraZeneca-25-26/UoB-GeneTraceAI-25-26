# Data Preview

```python
def preview(df, name):
    """
    Print a summary preview of a DataFrame including shape, columns, dtypes, 
    missing values percentage, and display first 5 rows.
    
    Parameters
    ----------
    df : pd.DataFrame
        DataFrame to preview
    name : str
        Name/label for the DataFrame in the output
    """
    print(f"{name}")
    print(f"Shape : {df.shape[0]:,} rows x {df.shape[1]:,} cols")
    print(f"Columns ({len(df.columns)}): {list(df.columns[:10])}" + (" ..." if len(df.columns)>10 else ""))
    print(f"Dtypes :\n{df.dtypes.value_counts().to_string()}")
    null_pct = (df.isnull().sum().sum() / df.size * 100)
    print(f"Missing: {null_pct:.1f}% of all values")
    print(f"\n Top 5 rows")
    display(df.head(5))
```

## Summary of the data sets

```python
# Cell 18 — summary table of all datasets
all_dfs = {
    1:  ("HPA RNA","hpa_rna",hpa_rna),
    2:  ("DepMap Expr","depmap_expr",depmap_expr),
    3:  ("GEO Expr","geo_expr",geo_expr),
    4:  ("Proteomics","proteomics",proteomics),
    5:  ("Fusions","fusions",fusions),
    6:  ("Mutations",       "mutations",        mutations),
    7:  ("Cellosaurus",     "cellosaurus",      cellosaurus),
    8:  ("DepMap Profiles", "depmap_profiles",  depmap_profiles),
    9:  ("Sample Info",     "sample_info",      sample_info),
    10: ("GEO Info",        "geo_info",         geo_info),
    11: ("HPA Desc",        "hpa_desc",         hpa_desc),
    12: ("Metabolomics",    "metabolomics",     metabolomics),
    13: ("miRNA",           "mirna",            mirna),
    14: ("Signatures",      "signatures",       signatures)}

rows = []
for num, (name, obj, df) in all_dfs.items():
    total_cells = df.size
    missing_cells = df.isnull().sum().sum()
    missing_pct = missing_cells / total_cells * 100 if total_cells > 0 else 0

    # column with most nulls
    worst_col = df.isnull().sum().idxmax()
    worst_pct  = df.isnull().sum().max() / len(df) * 100

    rows.append({
        "#":            num,
        "File":         name,
        "Object":       obj,
        "Rows":         f"{df.shape[0]:,}",
        "Cols":         f"{df.shape[1]:,}",
        "Missing cells":f"{missing_cells:,}",
        "Missing %":    f"{missing_pct:.1f}%",
        "Worst col":    worst_col,
        "Worst col %":  f"{worst_pct:.1f}%"})

summary = pd.DataFrame(rows)

# colour-code the Missing % column: green → yellow → red
def colour_missing(val):
    pct = float(val.strip('%'))
    if pct == 0:
        return 'background-color: #d4edda; color: #155724'
    elif pct < 5:
        return 'background-color: #fff3cd; color: #856404'
    elif pct < 20:
        return 'background-color: #ffd699; color: #6d3a00'
    else:
        return 'background-color: #f8d7da; color: #721c24'

summary.style \
    .map(colour_missing, subset=["Missing %", "Worst col %"]) \
    .set_properties(**{'text-align': 'left'}) \
    .hide(axis='index')
```

# Converting all Files to Parquet

```python
# Cell 19 — convert all 14 files to Parquet
import pandas as pd
import os

BASE = "/Users/trivenidhamdhere/Documents/MSc Data Science/MSc Dissertation - AZ/data"  # your root folder — change if needed
OUT  = os.path.join(BASE, "parquet")
os.makedirs(OUT, exist_ok=True)

files = [
    (1,"gene expression","1_4_hpa_rna_celline.tsv","\t",0,None),
    (2,"gene expression","2_DepMap_OmicsExpressionAllGenesTPMLogp1Profile.csv",",",0,0),
    (3,"gene expression","3_GEOexpression.csv","\t",0,None),
    (4,"gene expression","4_Harmonized_MS_CCLE_Gygi_subsetted.csv",",",0,None),
    (5,"gene properties","5_OmicsFusionFilteredSupplementary.csv",",",0,None),
    (6,"gene properties","6_OmicsSomaticMutationsProfile.csv",",",0,None),
    (7,"nomenclature","7_cellosaurus.csv",",",0,None),
    (8,"nomenclature","8_DepMap_OmicsProfiles.csv",",",0,None),
    (9,"nomenclature","9_DepMap_sample_info.csv",",",0,None),
    (10,"nomenclature","10_GEOInfo.txt","\t",0,None),
    (11,"nomenclature", "11_hpa_rna_celline_description.tsv","\t",0,None),
    (12,"non gene expression","12_CCLE_metabolomics_20190502.csv",",",0,None),
    (13,"non gene expression","13_CCLE_miRNA_20181103.gct","\t",2,None),
    (14,"non gene expression","14_OmicsGlobalSignatures.csv",",",0,None)]

results = []
for num, folder, fname, sep, skiprows, index_col in files:
    src = os.path.join(BASE, folder, fname)
    stem = fname.rsplit(".", 1)[0]          # filename without extension
    out_path = os.path.join(OUT, f"{stem}.parquet")

    try:
        df = pd.read_csv(
            src,
            sep=sep,
            skiprows=skiprows if skiprows > 0 else None,
            index_col=index_col,
            low_memory=False)

        # file 3 (GEO) — fall back to comma if tab gives 1 column
        if num == 3 and df.shape[1] == 1:
            df = pd.read_csv(src, low_memory=False)

        df.to_parquet(out_path, index=(index_col is not None), engine="fastparquet")

        size_mb = os.path.getsize(out_path) / 1_048_576
        results.append({
            "#": num,
            "File": fname,
            "Shape": f"{df.shape[0]:,} × {df.shape[1]:,}",
            "Parquet size": f"{size_mb:.1f} MB"})

    except Exception as e:
        results.append({
            "#": num,
            "File": fname,
            "Shape": "—",
            "Parquet size": "—",
            "Status": f" {e}",
        })
        print(f"[{num:02d}]  {fname}  →   {e}")

display(pd.DataFrame(results))
```
